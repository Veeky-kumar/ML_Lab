```python
hrs = [0, 1, 2, 3, 3, 5,5,5, 6, 7, 7, 10]
marks = [96, 85, 82, 74, 95, 68, 76, 84, 58, 65, 75, 50]
total_hrs = sum(hrs)
total_marks = sum(marks)
product_sum_of_hrs_and_marks = 0
for i in range(len(hrs)):
    product_sum_of_hrs_and_marks += hrs[i] * marks[i]

sum_of_squares_of_hrs = 0
for i in range(len(hrs)):
    sum_of_squares_of_hrs += hrs[i] ** 2
square_of_sum_of_hrs = total_hrs ** 2
n = len(hrs)

m = (n * product_sum_of_hrs_and_marks - total_hrs * total_marks) / (n * sum_of_squares_of_hrs - square_of_sum_of_hrs)
print(f"slope = {m:.2f}")
c = ((total_marks/12 )- m * (total_hrs/12)) 
print(f"for 9 hours of study, marks = {m*9 + c:.2f}")
```
